```python
import mlflow
import pandas as pd
from dotenv import load_dotenv

load_dotenv()
```

```python
df_test = pd.read_csv("./data/test.csv").head(10)

# preprocess
df_test = df_test.drop(columns=["product_id", "type", "target"])
df_test.head()
```

```python
MODEL_NAME = "workspace.default.predictive_maintenance"
MODEL_VERSION = "1"

mlflow.set_tracking_uri("databricks")
model = mlflow.pyfunc.load_model(f"models:/{MODEL_NAME}/{MODEL_VERSION}")
predictions = model.predict(df_test)
print(predictions)
```
